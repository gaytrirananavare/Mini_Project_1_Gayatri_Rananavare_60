```python
## # Mini Project: Web Browser Back Button Using Stack

class WebBrowser:
    def __init__(self):
        self.history = []
        self.current_page = None

    def visit_page(self, url):
        if self.current_page is not None:
            self.history.append(self.current_page)

        self.current_page = url
        print("Page opened:", self.current_page)

    def go_back(self):
        if self.history:
            self.current_page = self.history.pop()
            print("Returned to:", self.current_page)
        else:
            print("No previous page available!")

    def show_current_page(self):
        if self.current_page:
            print("Current Page:", self.current_page)
        else:
            print("No page is currently open.")

    def show_history(self):
        if self.history:
            print("\nBrowsing History:")
            for page in reversed(self.history):
                print(page)
        else:
            print("No browsing history available.")


browser = WebBrowser()

while True:
    print("\n===== WEB BROWSER =====")
    print("1. Visit New Page")
    print("2. Back")
    print("3. Show Current Page")
    print("4. Show Browsing History")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        url = input("Enter website URL: ")
        browser.visit_page(url)

    elif choice == '2':
        browser.go_back()

    elif choice == '3':
        browser.show_current_page()

    elif choice == '4':
        browser.show_history()

    elif choice == '5':
        print("Exiting Web Browser...")
        break

    else:
        print("Invalid choice! Please try again.")

```
